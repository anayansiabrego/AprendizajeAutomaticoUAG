```python
# Inciso 1
# Swaping Web. Ejecutar el siguiente codigo. Mandar snapshot de la salida del programa, con una explicación de lo que esta haciendo el código.
import requests
from bs4 import BeautifulSoup

# URL de la página web que queremos extraer
url = 'https://example.com'

# Realizar la solicitud a la página web
response = requests.get(url)

# Crear objeto BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')

print(response.text)

# Extraer datos específicos (por ejemplo, todos los # encabezados h2, h1, head, title, etc.)
headings = soup.find_all('h1')
i=1
for heading in headings:
    print("Result: ", i)
    print(heading.text)
```

Se realiza una petición a la url dada ( https://example.com) y se guarda la respuesta de la página web. 

Esa respuesta se pasa como argumento en la función BeautifulSoup, para convertir el texto de la página web en un árbol ordenado de datos. 
La respuesta dada en la petición se muestra de manera completa y sin ningún procesamiento a través de la instrucción print.  
Del arbol creado a través de soup, se hace una búsqueda y filtrado de datos, en este caso de todos los encabezados tipo h1, los cuales a través de un ciclo for se imprimen.


```python
# Inciso 2
# Swaping json. Ejecutar el siguiente codigo. Mandar snapshot de la salida del programa, con una explicación de lo que esta haciendo el código.
import requests

# URL de la API de GitHub
url = 'https://api.github.com/repos/python/cpython'

# Realizar la solicitud a la API
response = requests.get(url)
data = response.json()

# Mostrar información específica del repositorio
print(f"Repository: {data['name']}")
print(f"Description: {data['description']}")
print(f"Stars: {data['stargazers_count']}")
print(f"Forks: {data['forks_count']}")
```

Una vez guardado el contenido, se imprimen los valores de las keys deseadas:  name, description, stargazers_count y forks_count. 


```python
# Inciso 4
#  Pandas. Ejecutar el siguiente codigo y mandar el archivo CSV resultante. Para esta practica necesitan el archivo cars.csv adjunto. Mandar snapshot de la salida del programa, con una explicación de lo que esta haciendo el código.

import pandas as pd
import numpy as np

# 1. Load the Dataset
df = pd.read_csv('./dataset/cars.csv')

# 2. Inspect the Dataset
print(df.head())
print(df.dtypes)

# 3. Handle Missing Values
df['hp'].fillna(df['hp'].mean(), inplace=True)
df.dropna(subset=['hp'], inplace=True)

# 4. Filter Data
df_filtered = df[df['cyl'] == 6]
print(df_filtered)

# 5. Transform Columns
def categorize_hp(hp):
    if hp > 150:
        return 'High'
    elif hp > 80:
        return 'Medium'
    else:
        return 'Low'

df_filtered['HP_Category'] = df_filtered['hp'].apply(categorize_hp)

# 6. Save the Cleaned Dataset
df_filtered.to_csv('./dataset/cleaned_cars_dataset.csv', index=False)

print("Data cleaning and preparation complete. The cleaned dataset is saved as 'cleaned_cars_dataset.csv'.")
```

Explicación. 


Este código carga datos desde un archivo con formato csv y lo guarda en una estructura tipo DataFrame de Pnadas, 
Se hace una inspección inicial de los datos mostrando las primeras 5 filas (df.head()) y el tipo de dato de cada una de las columnas df.dtypes()

En los datos originales existen valores faltantes en la columna hp, los cuales se rellenan utilizando la media de los datos de esa columna, en caso que haya quedado alguna fila con algún valor faltante, se elimina. 

En un nuevo dataframe se guardan los datos de todos los autos de 6 cilindros, es decir, aquellos que en la columna cyl tengan un valor de 6 .
Crea una nueva columna en el dataFrame df_filtered llamada HP$\_$Category, en la cual se le asigna el valor High, Medium o Low, dependiendo de su valor en la columna HP

Finalmente, el dataframe resultante se almacena en un archivo csv e imprime el mensaje Data cleaning and preparation complete. The cleaned dataset is saved as 'cleaned_cars_dataset.csv'.
